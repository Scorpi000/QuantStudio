```python
import warnings
warnings.filterwarnings('ignore')
import datetime as dt

from QuantStudio.Tools.Visualization import qs_help
```

注意：运行下面的代码首先要执行 [gen_demo_factor_data.py](../tools/gen_demo_factor_data.py) 脚本生成示例数据。

# HDF5DB

![HDF5 因子库](https://gitee.com/scorpi000/QSImage/raw/master/QuantStudio/HDF5因子库.png)

```python
# 创建因子库对象并 connect
from QuantStudio.Factor.HDF5DB import HDF5DB

HDB = HDF5DB(args={"MainDir": "../data/HDF5"}).connect()
print(qs_help(HDB))
```

## 因子表

```python
# 因子库中的因子表列表
print(HDB.TableNames)
```

```python
# 获取因子表对象
FT = HDB.getTable("stock_cn_day_bar", args={"LookBack": 4})
print(qs_help(FT))
```

```python
# 因子表读取数据
DTs = [dt.datetime(2025, 1, 1) + dt.timedelta(i) for i in range(5)]
IDs = ["000001.SZ", "000002.SZ"]

Data = FT.readData(factor_names=["open", "close"], ids=IDs, dts=DTs)
print("因子表数据")
print(Data)
```

## 因子

```python
# 因子表中的因子列表
print(f"因子表 '{FT.Name}' 中的因子 : ", FT.FactorNames)
```

```python
# 获取因子对象
F = FT.getFactor("close")
print(qs_help(F))
```

```python
# 因子读取数据
DTs = [dt.datetime(2025, 1, 1) + dt.timedelta(i) for i in range(5)]
IDs = ["000001.SZ", "000002.SZ"]

Data = F.readData(ids=IDs, dts=DTs)
print("因子数据")
print(Data)
```
